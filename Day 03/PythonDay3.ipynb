Digit counter

```python
n = int(input())
count = 0
for i in range(n):
  count += 1
  n //= 10
  if(n <= 0):
    break
print(count)
```

Sum of Digits

```python
n = int(input())
sum = 0
for i in range(n):
  digit = n%10
  sum += digit
  n //=10
print(sum)
```

Palindrom

```python
str = input()
if(str == str[::-1]):
  print("True")
else:
  print("False")
```

```python
n = int(input())
num = n
r = 0
while n>0:
  d = n%10
  r *= 10
  r += d
  n //= 10
if num == r:
  print("True")
else:
  print("False")
```

Reversing a Number


```python
n  = int(input())
r = 0
while n>0:
  d = n%10
  r *= 10
  r += d
  n //= 10
print(r)
```

Printing Primes

```python
n = int(input())
for i in range(2,n+1):
  flag = 0
  for j in range(2,i):
    if(i%j==0):
      flag = 1
      break
  if(flag==0):
    print(i)
```
